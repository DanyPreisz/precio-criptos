# Análisis de Precios de Criptomonedas

Proyecto de consumo de API listo para **Google Colab**.

**Pregunta:** ¿cómo obtener y analizar precios de criptos?

**Método:** requests a CoinGecko API

**Stack:** requests, pandas, matplotlib

Datos en tiempo real.

## 0. Setup

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
print("OK")
```

## 1. Obtener precios (CoinGecko)

```python
url = "https://api.coingecko.com/api/v3/coins/markets"
params = {
    "vs_currency": "usd",
    "order": : "market_cap_desc",
    "per_page": : 10,
    "page": : 1
}
response = requests.get(url, params=params)
data = response.json()
print(f"Cryptos obtenidas: {len(data)}")
```

## 2. Procesar datos

```python
df = pd.DataFrame(data)
df = df[["id", "symbol", "current_price", "market_cap", "price_change_percentage_24h"]]
print(df.head())
```

## 3. Visualización

```python
plt.figure(figsize=(8, 4))
plt.bar(df["symbol"], df["current_price"])
plt.title("Precio actual (USD)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.bar(df["symbol"], df["price_change_percentage_24h"])
plt.title("Cambio % en 24h")
plt.axhline(0, color="red", linestyle="--")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 4. Conclusiones

- CoinGecko API es gratuita y no requiere key para endpoints básicos.
- Ideal para dashboards de criptos o análisis de mercado.
- Se puede extender a series históricas o predicción.
